```python
import xml.etree.ElementTree as ET
import json
enums_path = r"fix_repository\FIX.4.4\Base\Enums.xml"
root = ET.parse(enums_path)

enums = {}
for enum in root.findall('Enum'):
    key = enum.find('Tag').text

    if key not in enums:
        enums[key] = {enum.find('Value').text: enum.find('SymbolicName').text}
    else:
        enums[key][enum.find('Value').text] = enum.find('SymbolicName').text



enums




```

```python
num1 = 120
num2 = 130
ans = 0
for num in range(num1, num2+1):
    if num < 100:
        continue
    num = [int(n) for n in str(num)]
    for i in range(1, len(num)-1):
        if num[i-1] < num[i] and num[i+1] < num[i]:
            ans += 1
            break
        if  num[i-1] > num[i] and num[i+1] > num[i]:
            ans += 1
            break

print(ans)
```

```python
s = "acedc"

ans = []
for i in range(len(s)):
    seen = {s[i]: 1}
    curr = [s[i]]
    for j in range(i,len(s)):
        if s[j] not in seen:
            seen[s[j]] = 1
        else:
            seen[s[j]] += 1
        
        if seen[s[j]] > 2:
            break
        else:
            curr.append(s[j])

    if len(curr) > len(ans):
        ans = curr

ans
```
